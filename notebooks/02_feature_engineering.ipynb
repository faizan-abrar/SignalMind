# **Import Libraries & Load Data**

```python
import pandas as pd

df = pd.read_csv("/content/ambient_temperature_system_failure.csv")
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.head()
```

# **Rolling Average Feature**

```python
df['rolling_avg_24h'] = df['value'].rolling(window=24).mean()
df.head(30)
```

# **Rolling Standard Deviation Feature**

```python
df['rolling_std_24h'] = df['value'].rolling(window=24).std()
df.head(30)
```

# **Lag Feature**

```python
df['lag_1h'] = df['value'].shift(1)
df.head(10)
```

# **Difference Feature (Change from Last Hour)**

```python
df['diff_1h'] = df['value'] - df['lag_1h']
df.head(10)
```

```python
df_clean = df.dropna()
df_clean.shape
```

# **Processed Data Save Karo**

```python
df_clean.to_csv("processed_temperature_data.csv", index=False)
```
