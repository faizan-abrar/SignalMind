# **Exploratory Data Analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/ambient_temperature_system_failure.csv")
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.head()
```

```python
df[df['value'] > 80]
```

# **Rolling Average**

```python
df['rolling_avg'] = df['value'].rolling(window=24).mean()

plt.figure(figsize=(15, 5))
plt.plot(df['timestamp'], df['value'], alpha=0.4, label='Original')
plt.plot(df['timestamp'], df['rolling_avg'], color='red', label='24-hour Rolling Average')
plt.title('Temperature with Rolling Average')
plt.xlabel('Time')
plt.ylabel('Temperature')
plt.legend()
plt.show()
```
